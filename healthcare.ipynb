```python
# 1. Female patients with Diabetes
df[(df["Gender"] == "Female") & (df["Medical Condition"] == "Diabetes")]

# 2. Age < 30 and Elective admission
df[(df["Age"] < 30) & (df["Admission Type"] == "Elective")]

# 3. High Billing Amount > 40000 and Male
df[(df["Billing Amount"] > 40000) & (df["Gender"] == "Male")]

# 4. Cancer patients with A+ Blood Type
df[(df["Medical Condition"] == "Cancer") & (df["Blood Type"] == "A+")]

# 5. 3 conditions - Age > 50, Female, Insurance = Medicare
df[
    (df["Age"] > 50) & 
    (df["Gender"] == "Female") & 
    (df["Insurance Provider"] == "Medicare")
]

```
