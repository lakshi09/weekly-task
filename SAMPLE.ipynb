```python
df["Order Date"] = pd.to_datetime(df["Order Date"])
df["Ship Date"] = pd.to_datetime(df["Ship Date"]) 
df[["Order Date", "Ship Date"]].dtypes
```

```python
text_columns = ["Category","Sub-Category","Segment"]

for column in text_columns:
    df[column]=(
        df[column]
        .astype(str)
        .str.strip()
        .str.title()
    )
```

```python
numerical_features = [
    "Sales",
    "Quantity",
    "Discount",
    "Profit",
]

df[numerical_features].describe()
```

```python
summary = pd.DataFrame({
    "Mean": df[numerical_features].mean(),
    "Median":df[numerical_features].median(),
    "Minimum":df[numerical_features].min(),
    "Maximum":df[numerical_features].max(),
    "Standard Deviation":df[numerical_features].std()
})
```

```python
category_sales = (
    df.groupby("Category")["Sales"]
    .sum()
    .sort_values(ascending=False)
)

category_sales
```

```python
region_profit = (
    df.groupby("Region")["Profit"]
    .sum()
    .sort_values(ascending=False)
)

region_profit
```
